## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">1 · Preparar el entorno y cargar los datos</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
La serie contiene 24 observaciones semestrales, de 2015-1 a 2026-2. Cada fila conserva dos elementos inseparables: **el valor observado** y **el periodo en que ocurrió**. Separarlos destruye la serie: un conteo sin fecha es un número suelto, no una observación temporal.

Todo el cálculo se apoya en pandas [2] y todas las gráficas en Matplotlib [3].

```python
# ─────────────────────────────────────────────────────────────────────────
# Laboratorio 1 · MIAAD · UACJ
# Entorno, identidad visual institucional y carga de la serie
# ─────────────────────────────────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

# --- Paleta institucional UACJ -------------------------------------------
# azul y oro se tomaron del escudo; serie_1..serie_3 son los pasos usados en
# las gráficas, elegidos para conservar contraste sobre fondo blanco.
UACJ = {
    "azul":      "#003DA6",   # azul del escudo, títulos y acentos
    "oro":       "#FFEA00",   # oro del escudo, subrayados
    "serie_1":   "#1157CC",   # serie observada / valor real
    "serie_2":   "#B87E00",   # baseline estacional
    "serie_3":   "#A8408A",   # baseline naïve
    "serie_4":   "#1B7A3E",   # baseline estacional con deriva (anexo A)
    "alerta":    "#C0392B",   # subestimación y avisos
    "tinta":     "#1F2933",
    "tinta_sec": "#5A6572",
    "rejilla":   "#DCE1E7",
    "suave":     "#F2F7FE",
}

# Un color por entidad, no por posición: el mismo método conserva su color
# en todas las gráficas del cuaderno.
COLOR = {
    "Real":       UACJ["serie_1"],
    "Naive":      UACJ["serie_3"],
    "Estacional": UACJ["serie_2"],
    "Deriva":     UACJ["serie_4"],
}

# El nombre que se imprime en gráficas no es la clave del diccionario.
NOMBRE = {
    "Real":       "Real",
    "Naive":      "Naïve",
    "Estacional": "Estacional",
    "Deriva":     "Estacional + deriva",
}

mpl.rcParams.update({
    "figure.figsize":      (10.6, 4.4),
    "figure.dpi":          105,
    "figure.facecolor":    "#FFFFFF",
    "savefig.facecolor":   "#FFFFFF",
    "axes.facecolor":      "#FFFFFF",
    "axes.edgecolor":      UACJ["rejilla"],
    "axes.linewidth":      1.0,
    "axes.labelcolor":     UACJ["tinta_sec"],
    "axes.titlecolor":     UACJ["azul"],
    "axes.titlesize":      13,
    "axes.titleweight":    "bold",
    "axes.titlelocation":  "left",
    "axes.titlepad":       12,
    "axes.labelsize":      10,
    "axes.spines.top":     False,
    "axes.spines.right":   False,
    "axes.grid":           True,
    "axes.axisbelow":      True,
    "grid.color":          UACJ["rejilla"],
    "grid.linewidth":      0.8,
    "text.color":          UACJ["tinta"],
    "xtick.color":         UACJ["tinta_sec"],
    "ytick.color":         UACJ["tinta_sec"],
    "xtick.labelsize":     9,
    "ytick.labelsize":     9,
    "legend.frameon":      False,
    "legend.fontsize":     9.5,
    "lines.linewidth":     2.2,
    "lines.markersize":    6,
    "font.family":         "sans-serif",
    "font.sans-serif":     ["DejaVu Sans", "Helvetica", "Arial"],
    "font.size":           10,
})

def titulo(ax, principal, secundario=None):
    """Título en azul institucional con una línea de contexto en tinta secundaria."""
    ax.set_title(principal, loc="left", color=UACJ["azul"], fontsize=12.5,
                 fontweight="bold", pad=26 if secundario else 12)
    if secundario:
        ax.text(0, 1.035, secundario, transform=ax.transAxes, fontsize=9.2,
                color=UACJ["tinta_sec"], ha="left", va="bottom")

def eje_anual(ax, marco):
    """Etiqueta el eje horizontal con el año, marcado en cada semestre 1."""
    s1 = marco[marco["Semestre"] == 1]
    ax.set_xticks(s1["t"].to_numpy())
    ax.set_xticklabels(s1["Periodo"].str[:4].to_numpy())
    ax.set_xticks(marco["t"].to_numpy(), minor=True)
    ax.set_xlabel("Año · la marca corresponde al semestre 1")

def rotula(ax, x, y, texto, color, dx=0.25, dy=0.0, ha="left"):
    """Etiqueta directa junto a un punto, con halo blanco para que no se pierda."""
    t = ax.text(x + dx, y + dy, texto, color=color, fontsize=9.5,
                fontweight="bold", ha=ha, va="center")
    t.set_path_effects([pe.withStroke(linewidth=3.2, foreground="#FFFFFF")])
    return t

def nota(fig, texto, alto=0.055):
    """Cierra la figura: reserva una franja inferior y escribe ahí la fuente.

    El pie va dentro del lienzo, no en coordenadas negativas: así sobrevive a un
    savefig normal y no depende de que el backend recorte con bbox_inches.
    """
    fig.tight_layout(rect=(0, alto, 1, 1))
    fig.text(0.005, alto / 2, texto, fontsize=8.2, color=UACJ["tinta_sec"],
             ha="left", va="center")

print(f"Entorno listo · pandas {pd.__version__} · numpy {np.__version__} · matplotlib {mpl.__version__}")
print(f"Paleta institucional cargada: {len(UACJ)} colores · azul {UACJ['azul']} y oro {UACJ['oro']} del escudo")
```

Con el entorno preparado, cargamos la serie. Son 24 conteos semestrales; el marco de datos añade solo dos columnas auxiliares, `t` para numerar las observaciones y `Semestre` para distinguir los periodos 1 y 2.

```python
# --- Serie de trabajo -----------------------------------------------------
periodos = [f"{anio}-{semestre}" for anio in range(2015, 2027) for semestre in (1, 2)]
deserciones = [
    41, 45, 43, 47, 44, 49, 46, 50,
    48, 53, 50, 55, 52, 56, 54, 59,
    56, 61, 58, 64, 60, 66, 62, 69
]

df = pd.DataFrame({"Periodo": periodos, "Deserciones": deserciones})
df["t"] = np.arange(1, len(df) + 1)
df["Semestre"] = df["Periodo"].str[-1].astype(int)

print(f"Serie cargada: {len(df)} observaciones semestrales, de {df.iloc[0]['Periodo']} a {df.iloc[-1]['Periodo']}\n")
print(df.to_string(index=False))
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">2 · Verificar la calidad y el orden temporal</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Antes de analizar patrones hay que revisar que los datos estén completos, sin duplicados y en el orden esperado. Un modelo no corrige automáticamente una serie mal construida: la reproduce con más confianza de la que merece.

```python
revisiones = {
    "Número de observaciones":            len(df),
    "Periodos duplicados":                int(df["Periodo"].duplicated().sum()),
    "Valores faltantes":                  int(df["Deserciones"].isna().sum()),
    "Primer periodo":                     df.iloc[0]["Periodo"],
    "Último periodo":                     df.iloc[-1]["Periodo"],
    "Todos los valores son no negativos": bool((df["Deserciones"] >= 0).all()),
    "Frecuencia constante (2 por año)":   bool(df.groupby(df["Periodo"].str[:4]).size().eq(2).all()),
    "Periodos en orden cronológico":      bool(df["Periodo"].is_monotonic_increasing),
}

ancho = max(len(k) for k in revisiones)
for clave, valor in revisiones.items():
    print(f"{clave:<{ancho}} : {valor}")

huecos = set(periodos) - set(df["Periodo"])
print(f"\n{'Huecos en el calendario':<{ancho}} : {'ninguno' if not huecos else sorted(huecos)}")
print(f"{'Rango observado':<{ancho}} : {df['Deserciones'].min()} a {df['Deserciones'].max()} estudiantes")
```

Esta separación no es aleatoria: respeta la dirección del tiempo. Bergmeir y Benítez [7] documentan por qué una partición aleatoria distorsiona la evaluación cuando las observaciones están ordenadas en el tiempo.

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">7 · Baseline 1: pronóstico naïve</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
El baseline naïve supone que el siguiente valor será igual al último observado:

$$\hat{y}_{t} = y_{t-1}$$

No pretende ser sofisticado. Su función es fijar un mínimo razonable que cualquier método más complejo debería superar: si un modelo elaborado no le gana al naïve, el modelo no está aportando información, solo costo. El naïve y el estacional son dos de los métodos de referencia que presenta el capítulo 5.2 de [6].

```python
df["Pronostico_naive"] = df["Deserciones"].shift(1)

tabla_naive = df.loc[test.index, ["Periodo", "Deserciones", "Pronostico_naive"]].copy()
tabla_naive.columns = ["Periodo", "Real", "Naive"]
tabla_naive["Origen"] = df.loc[test.index - 1, "Periodo"].to_numpy()
tabla_naive = tabla_naive[["Origen", "Periodo", "Real", "Naive"]]

print(tabla_naive.to_string(index=False))
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">8 · Baseline 2: pronóstico estacional</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Como la gráfica y las medias por semestre mostraron una alternancia estable entre los semestres 1 y 2, usaremos el mismo semestre del año anterior:

$$\hat{y}_{t} = y_{t-2}$$

El rezago 2 tiene sentido porque la frecuencia es semestral: dos observaciones equivalen a un año, de modo que $y_{t-2}$ es el periodo comparable del calendario. En una serie mensual el rezago análogo sería 12; en una trimestral, 4.

```python
df["Pronostico_estacional"] = df["Deserciones"].shift(2)

tabla_baselines = df.loc[test.index, [
    "Periodo", "Deserciones", "Pronostico_naive", "Pronostico_estacional"
]].copy()
tabla_baselines.columns = ["Periodo", "Real", "Naive", "Estacional"]
print(tabla_baselines.to_string(index=False))

fig, ax = plt.subplots(figsize=(9.6, 4.3))
x = np.arange(len(tabla_baselines))

ax.plot(x, tabla_baselines["Real"], marker="o", linewidth=3.0, color=COLOR["Real"],
        markerfacecolor="#FFFFFF", markeredgecolor=COLOR["Real"], markeredgewidth=2.0,
        label="Real", zorder=4)
ax.plot(x, tabla_baselines["Naive"], marker="s", linestyle="--", linewidth=2.0,
        color=COLOR["Naive"], label="Naïve", zorder=3)
ax.plot(x, tabla_baselines["Estacional"], marker="^", linestyle="--", linewidth=2.0,
        color=COLOR["Estacional"], label="Estacional", zorder=3)

for nombre in ["Real", "Naive", "Estacional"]:
    valor = tabla_baselines[nombre].iloc[-1]
    rotula(ax, x[-1], valor, f" {NOMBRE[nombre]} · {valor:.0f}", COLOR[nombre], dx=0.06)

ax.set_xticks(x)
ax.set_xticklabels(tabla_baselines["Periodo"])
ax.set_xlim(-0.25, len(x) - 0.05)
ax.set_ylim(54, 74)
ax.set_yticks(np.arange(54, 75, 2))
ax.set_ylabel("Estudiantes por semestre")
ax.set_xlabel("Periodo pronosticado")
ax.legend(loc="upper left", ncols=3)
titulo(ax, "Valores reales y baselines en el conjunto de prueba",
       "Marcador distinto por método: la identidad no depende solo del color")

nota(fig, "Fuente: datos simulados del laboratorio. Cada punto pronosticado usa exclusivamente información "
          "anterior a su propio periodo.")
plt.show()
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">9 · Calcular el error periodo por periodo</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Definimos el error de pronóstico como la diferencia entre lo que ocurrió y lo que se había anticipado:

$$e_t = y_t - \hat{y}_t$$

- Si `error > 0`, el método **subestimó**: el valor real fue mayor que el pronóstico.
- Si `error < 0`, el método **sobreestimó**: el pronóstico fue mayor que el valor real.
- El error absoluto $|e_t|$ expresa cuánto se equivocó, sin considerar la dirección.

```python
resultados = tabla_baselines.copy()
resultados["Error_naive"]      = resultados["Real"] - resultados["Naive"]
resultados["Abs_naive"]        = resultados["Error_naive"].abs()
resultados["Error_estacional"] = resultados["Real"] - resultados["Estacional"]
resultados["Abs_estacional"]   = resultados["Error_estacional"].abs()

print(resultados.to_string(index=False))
print(f"\nSuma de errores con signo · naïve      : {resultados['Error_naive'].sum():+.0f}  "
      f"(sesgo medio {resultados['Error_naive'].mean():+.2f})")
print(f"Suma de errores con signo · estacional : {resultados['Error_estacional'].sum():+.0f}  "
      f"(sesgo medio {resultados['Error_estacional'].mean():+.2f})")

fig, ejes = plt.subplots(1, 2, figsize=(11.4, 4.2), sharey=True)
limite = 8.4

for ax, metodo, columna in zip(ejes, ["Naive", "Estacional"], ["Error_naive", "Error_estacional"]):
    errores = resultados[columna].to_numpy()
    colores = [UACJ["alerta"] if e > 0 else COLOR["Real"] for e in errores]
    ax.bar(resultados["Periodo"], errores, color=colores, width=0.56, zorder=3)
    for periodo, e in zip(resultados["Periodo"], errores):
        ax.text(periodo, e + (0.55 if e > 0 else -0.55), f"{e:+.0f}",
                ha="center", va="bottom" if e > 0 else "top",
                fontsize=10, fontweight="bold",
                color=UACJ["alerta"] if e > 0 else COLOR["Real"])
    ax.axhline(0, color=UACJ["tinta_sec"], linewidth=1.2, zorder=4)
    ax.set_ylim(-limite, limite)
    ax.grid(axis="x", visible=False)
    titulo(ax, f"{NOMBRE[metodo]} · MAE = {resultados['Abs_' + metodo.lower()].mean():.2f}",
           "error = real − pronóstico, en estudiantes")

ejes[0].set_ylabel("Error con signo")
ejes[0].text(-0.42, 6.9, "arriba de cero:\nSUBESTIMÓ", color=UACJ["alerta"], fontsize=9,
             fontweight="bold", ha="left", va="center")
ejes[0].text(-0.42, -6.9, "abajo de cero:\nSOBREESTIMÓ", color=COLOR["Real"], fontsize=9,
             fontweight="bold", ha="left", va="center")

nota(fig, "Fuente: datos simulados del laboratorio. Escala compartida entre ambos paneles para que las "
          "alturas sean comparables a simple vista.")
plt.show()
```

### <span style="color:#8A6100;font-weight:700;letter-spacing:-.2px;">Cómo interpretar el resultado</span>
<div style="border-left:5px solid #B87E00;background:#FFF9E8;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;">La tabla y la gráfica permiten distinguir dos preguntas que una métrica agregada confunde:</p><ul style="margin:0;padding-left:20px;"><li style="margin-bottom:6px;"><b>Dirección.</b> ¿El método subestima o sobreestima? El naïve alterna: sobreestima en los semestres 1 y subestima en los semestres 2, porque arrastra el semestre equivocado. El error del estacional, en cambio, es positivo en los cuatro periodos: subestima siempre.</li><li style="margin-bottom:6px;"><b>Magnitud.</b> ¿Por cuántos estudiantes se equivoca? Aquí el estacional se mantiene entre 2 y 3; el naïve llega a 7.</li></ul><p style="margin:0 0 10px 0;">Un promedio de errores absolutos resume la magnitud pero <b>oculta la dirección</b>: dos métodos con el mismo MAE pueden tener consecuencias muy distintas si uno falla siempre hacia el mismo lado. Por eso conviene conservar la tabla de errores con signo antes de calcular cualquier promedio, y por eso la suma de errores con signo (+5 en el naïve, +9 en el estacional) es un diagnóstico de sesgo, no una métrica de exactitud.</p></div></div>

### <span style="color:#1157CC;font-weight:700;letter-spacing:-.2px;">Respuesta del estudiante · 9</span>
<div style="border-left:5px solid #1157CC;background:#F2F7FE;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:11.5px;font-weight:700;letter-spacing:1.3px;text-transform:uppercase;color:#1157CC;margin-bottom:6px;">Consigna del laboratorio</div><div style="color:#5A6572;font-size:13.5px;font-style:italic;line-height:1.55;margin-bottom:11px;">Indique en qué periodos el naïve subestimó y en cuáles sobreestimó. ¿Qué patrón observa en el error estacional?</div><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;"><b>Naïve.</b> Sobreestimó en 2025-1 (error &minus;4) y en 2026-1 (error &minus;4), es decir, en los dos semestres 1 del conjunto de prueba, porque en ambos casos copió un semestre 2, que en esta serie siempre es más alto. Subestimó en 2025-2 (error +6) y en 2026-2 (error +7), los dos semestres 2, por la razón inversa. Su error alterna de signo al ritmo de la estacionalidad.</p><p style="margin:0 0 10px 0;"><b>Estacional.</b> El patrón es distinto y más informativo: los cuatro errores son positivos (+2, +2, +2 y +3), de modo que el método <b>subestima de forma sistemática</b>. No se trata de ruido, es un sesgo.</p><p style="margin:0 0 10px 0;"><b>Explicación.</b> La regla ŷ<sub>t</sub> = y<sub>t&minus;2</sub> reproduce el nivel del mismo semestre del año anterior, con lo que captura correctamente la estacionalidad, pero ignora por completo la tendencia. Como la serie crece unos dos estudiantes al año en cada semestre, el pronóstico llega sistemáticamente corto por esa misma cantidad. El sesgo del método coincide, de hecho, con el crecimiento anual de la serie, y esa coincidencia sugiere de inmediato la corrección: añadir la deriva. Esa idea se desarrolla en el anexo A.</p><p style="margin:0 0 10px 0;">Conviene subrayar la consecuencia para la decisión: un método que subestima siempre es más peligroso aquí que uno que se equivoca por igual en ambos sentidos, porque la subestimación es justo el error que deja estudiantes sin acompañamiento.</p></div></div>

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">10 · Resumir con MAE, RMSE y MAPE</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Calcularemos tres métricas complementarias:

$$\text{MAE}=\frac{1}{n}\sum_{t=1}^{n}\left|e_t\right| \qquad
\text{RMSE}=\sqrt{\frac{1}{n}\sum_{t=1}^{n}e_t^{2}} \qquad
\text{MAPE}=\frac{100}{n}\sum_{t=1}^{n}\left|\frac{e_t}{y_t}\right|$$

- **MAE:** error absoluto promedio, en estudiantes. Es la unidad de la decisión.
- **RMSE:** da mayor peso a los errores grandes y regresa a la unidad original.
- **MAPE:** error porcentual promedio; facilita comunicar proporciones, pero es inestable cerca de cero y penaliza de forma asimétrica las sobreestimaciones.

Ninguna métrica es universalmente mejor: se elige según la escala de la serie y el costo de equivocarse. Hyndman y Koehler [8] revisan las limitaciones de cada una y proponen el error escalado como alternativa al MAPE.

```python
def metricas(real, pronostico):
    """MAE, RMSE y MAPE de un vector de pronósticos frente al valor observado."""
    real       = np.asarray(real, dtype=float)
    pronostico = np.asarray(pronostico, dtype=float)
    error = real - pronostico
    mae   = np.mean(np.abs(error))
    rmse  = np.sqrt(np.mean(error ** 2))
    mape  = np.mean(np.abs(error / real)) * 100
    return mae, rmse, mape

filas_metricas = []
for nombre in ["Naive", "Estacional"]:
    mae, rmse, mape = metricas(resultados["Real"], resultados[nombre])
    filas_metricas.append({"Método": nombre, "MAE": mae, "RMSE": rmse, "MAPE_%": mape})

comparacion = pd.DataFrame(filas_metricas)
print(comparacion.round(2).to_string(index=False))

mejora = 1 - comparacion.loc[1, "MAE"] / comparacion.loc[0, "MAE"]
print(f"\nReducción del MAE al pasar del naïve al estacional: {mejora:.1%}")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.4, 4.3),
                               gridspec_kw={"width_ratios": [2, 1]})
x = np.arange(2)
ancho = 0.30

# Panel A · métricas en estudiantes. MAPE va aparte: sus unidades son otras.
for i, metodo in enumerate(["Naive", "Estacional"]):
    valores = comparacion.loc[comparacion["Método"] == metodo, ["MAE", "RMSE"]].to_numpy()[0]
    posicion = x + (i - 0.5) * (ancho + 0.02)
    barras = ax1.bar(posicion, valores, width=ancho, color=COLOR[metodo],
                     label="Naïve" if metodo == "Naive" else metodo, zorder=3)
    for barra, valor in zip(barras, valores):
        ax1.text(barra.get_x() + barra.get_width() / 2, valor + 0.13, f"{valor:.2f}",
                 ha="center", va="bottom", fontsize=10.5, fontweight="bold", color=COLOR[metodo])

ax1.set_xticks(x)
ax1.set_xticklabels(["MAE", "RMSE"])
ax1.set_xlim(-0.62, 1.62)
ax1.set_ylabel("Estudiantes")
ax1.set_ylim(0, 6.6)
ax1.grid(axis="x", visible=False)
ax1.legend(loc="upper right", ncols=2)
titulo(ax1, "A · Error medio en la unidad de la decisión", "Menor es mejor · conjunto de prueba, 4 periodos")

# Panel B · MAPE, en porcentaje
valores_mape = comparacion["MAPE_%"].to_numpy()
barras = ax2.bar(["Naïve", "Estacional"], valores_mape,
                 color=[COLOR["Naive"], COLOR["Estacional"]], width=0.55, zorder=3)
for barra, valor in zip(barras, valores_mape):
    ax2.text(barra.get_x() + barra.get_width() / 2, valor + 0.2, f"{valor:.2f} %",
             ha="center", va="bottom", fontsize=10.5, fontweight="bold", color=barra.get_facecolor())
ax2.set_ylabel("Error porcentual medio")
ax2.set_ylim(0, 10)
ax2.grid(axis="x", visible=False)
titulo(ax2, "B · MAPE", "Unidad distinta: no comparable con el panel A")

nota(fig, "Fuente: datos simulados del laboratorio. MAE y RMSE se grafican juntos porque comparten unidad; "
          "el MAPE se separa para no forzar dos escalas en un mismo eje.")
plt.show()
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">11 · Incorporar el costo de subestimar</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
MAE, RMSE y MAPE penalizan por igual un error de `+4` y uno de `-4`. El contexto, en cambio, no es simétrico: subestimar puede dejar estudiantes en riesgo sin atención, mientras que sobreestimar solo deja capacidad ociosa.

Usaremos un costo ilustrativo:

$$C(e_t) = 2\,\left|e_t\right| \quad \text{si } e_t > 0 \;\; (\text{subestimó})$$

$$C(e_t) = 1\,\left|e_t\right| \quad \text{si } e_t \le 0 \;\; (\text{sobreestimó})$$

Los pesos 2 y 1 **no son una verdad institucional**: son un supuesto explícito que sirve para mostrar cómo la decisión modifica la evaluación. En un proyecto real habría que acordarlos con quien responde por el proceso. Kolassa [9] muestra que el pronóstico puntual que conviene emitir cambia según la medida de error o de exactitud que se adopte.

```python
def costo_asimetrico(real, pronostico, peso_subestimar=2, peso_sobreestimar=1):
    """Costo por periodo cuando equivocarse hacia abajo es más caro que hacia arriba."""
    error = np.asarray(real, dtype=float) - np.asarray(pronostico, dtype=float)
    return np.where(error > 0,
                    peso_subestimar * np.abs(error),
                    peso_sobreestimar * np.abs(error))

for metodo in ["Naive", "Estacional"]:
    resultados[f"Costo_{metodo}"] = costo_asimetrico(resultados["Real"], resultados[metodo])

print(resultados[["Periodo", "Real",
                  "Naive", "Error_naive", "Costo_Naive",
                  "Estacional", "Error_estacional", "Costo_Estacional"]].to_string(index=False))

costo_naive      = resultados["Costo_Naive"].mean()
costo_estacional = resultados["Costo_Estacional"].mean()
print(f"\nCosto promedio · naïve      : {costo_naive:.2f} puntos")
print(f"Costo promedio · estacional : {costo_estacional:.2f} puntos")
print(f"Proporción del costo del estacional que proviene de subestimar: "
      f"{100 * resultados.loc[resultados['Error_estacional'] > 0, 'Costo_Estacional'].sum() / resultados['Costo_Estacional'].sum():.0f} %")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.4, 4.3),
                               gridspec_kw={"width_ratios": [1.25, 1]})

# Panel A · la forma de la función de costo
e_neg = np.linspace(-8, 0, 60)
e_pos = np.linspace(0, 8, 60)
ax1.plot(e_neg, np.abs(e_neg), linewidth=2.6, color=COLOR["Real"], label="Sobreestimar · peso 1")
ax1.plot(e_pos, 2 * np.abs(e_pos), linewidth=2.6, color=UACJ["alerta"], label="Subestimar · peso 2")
ax1.fill_between(e_pos, 0, 2 * np.abs(e_pos), color=UACJ["alerta"], alpha=0.08)
ax1.fill_between(e_neg, 0, np.abs(e_neg), color=COLOR["Real"], alpha=0.08)
ax1.axvline(0, color=UACJ["tinta_sec"], linewidth=1.2)
ax1.set_xlabel("Error con signo · estudiantes")
ax1.set_ylabel("Costo · puntos")
ax1.set_xlim(-8, 8)
ax1.set_ylim(0, 17)
ax1.legend(loc="upper center", ncols=1)
titulo(ax1, "A · La función de costo no es simétrica",
       "El mismo error absoluto cuesta el doble si el pronóstico se quedó corto")

# Panel B · costo promedio por método, con la parte que proviene de subestimar
metodos = ["Naïve", "Estacional"]
por_subestimar = [
    resultados.loc[resultados["Error_naive"] > 0, "Costo_Naive"].sum() / len(resultados),
    resultados.loc[resultados["Error_estacional"] > 0, "Costo_Estacional"].sum() / len(resultados),
]
por_sobreestimar = [costo_naive - por_subestimar[0], costo_estacional - por_subestimar[1]]

ax2.bar(metodos, por_subestimar, width=0.55, color=UACJ["alerta"],
        label="proviene de subestimar", zorder=3)
ax2.bar(metodos, por_sobreestimar, bottom=np.array(por_subestimar) + 0.06, width=0.55,
        color=COLOR["Real"], label="proviene de sobreestimar", zorder=3)
for i, total in enumerate([costo_naive, costo_estacional]):
    ax2.text(i, total + 0.28, f"{total:.2f}", ha="center", va="bottom",
             fontsize=11.5, fontweight="bold", color=UACJ["tinta"])
ax2.set_ylabel("Costo promedio · puntos")
ax2.set_ylim(0, 10.5)
ax2.grid(axis="x", visible=False)
ax2.legend(loc="upper right", ncols=1)
titulo(ax2, "B · Costo promedio y su composición",
       "Todo el costo del estacional viene de la dirección cara")

nota(fig, "Fuente: datos simulados del laboratorio. Los pesos 2 y 1 son un supuesto ilustrativo del "
          "ejercicio, no una cifra acordada institucionalmente.")
plt.show()
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">12 · Validación temporal rolling-origin</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Una sola separación puede favorecer accidentalmente a un método: basta que el corte caiga en un tramo que le convenga. La validación *rolling-origin*, u origen móvil, repite el ensayo en varios momentos:

1. Se fija un origen histórico.
2. Se utiliza únicamente la información anterior a él.
3. Se pronostica el siguiente periodo.
4. Se revela el valor real y el origen avanza un paso.

Así se imita la forma en que el pronóstico se usaría en la realidad, y se obtienen varias mediciones de error en lugar de una. Es el esquema de evaluación con origen móvil que revisa Tashman [10], y la misma disciplina que sostiene las comparaciones a gran escala como la M4, con 100 000 series y 61 métodos [11]; por eso conviene fijar siempre un piso simple antes de complicar el método.

```python
evaluaciones = []

# Comenzamos en 2022-1 para reunir diez orígenes recientes y consecutivos.
inicio = df.index[df["Periodo"] == "2022-1"][0]

for i in range(inicio, len(df)):
    real            = df.loc[i, "Deserciones"]
    pred_naive      = df.loc[i - 1, "Deserciones"]   # y(t-1)
    pred_estacional = df.loc[i - 2, "Deserciones"]   # y(t-2)
    evaluaciones.append({
        "Origen":               df.loc[i - 1, "Periodo"],
        "Periodo_pronosticado": df.loc[i, "Periodo"],
        "Real":                 real,
        "Naive":                pred_naive,
        "Estacional":           pred_estacional,
        "Abs_naive":            abs(real - pred_naive),
        "Abs_estacional":       abs(real - pred_estacional),
    })

rolling = pd.DataFrame(evaluaciones)
print(rolling.to_string(index=False))

mae_roll_naive      = rolling["Abs_naive"].mean()
mae_roll_estacional = rolling["Abs_estacional"].mean()
ganados = int((rolling["Abs_estacional"] < rolling["Abs_naive"]).sum())
empates = int((rolling["Abs_estacional"] == rolling["Abs_naive"]).sum())

print(f"\nOrígenes evaluados                  : {len(rolling)}")
print(f"MAE rolling-origin · naïve          : {mae_roll_naive:.2f} estudiantes")
print(f"MAE rolling-origin · estacional     : {mae_roll_estacional:.2f} estudiantes")
print(f"Orígenes en que ganó el estacional  : {ganados} de {len(rolling)}  (empates: {empates})")

fig, ax = plt.subplots(figsize=(10.8, 4.4))
x = np.arange(len(rolling))

ax.plot(x, rolling["Abs_naive"], marker="s", linewidth=2.2, color=COLOR["Naive"],
        label="Naïve · error absoluto", zorder=3)
ax.plot(x, rolling["Abs_estacional"], marker="^", linewidth=2.2, color=COLOR["Estacional"],
        label="Estacional · error absoluto", zorder=3)
ax.axhline(mae_roll_naive, linestyle="--", linewidth=1.4, color=COLOR["Naive"], alpha=0.65, zorder=2)
ax.axhline(mae_roll_estacional, linestyle="--", linewidth=1.4, color=COLOR["Estacional"], alpha=0.65, zorder=2)

rotula(ax, x[-1], mae_roll_naive, f" MAE {mae_roll_naive:.2f}", COLOR["Naive"], dx=0.12)
rotula(ax, x[-1], mae_roll_estacional, f" MAE {mae_roll_estacional:.2f}", COLOR["Estacional"], dx=0.12)

ax.set_xticks(x)
ax.set_xticklabels(rolling["Periodo_pronosticado"], rotation=45, ha="right")
ax.set_xlim(-0.3, len(x) + 0.9)
ax.set_ylim(0, 8.2)
ax.set_ylabel("Error absoluto · estudiantes")
ax.set_xlabel("Periodo pronosticado · un paso adelante desde el origen inmediato anterior")
ax.legend(loc="upper left", ncols=2)
titulo(ax, "Error absoluto en diez orígenes consecutivos",
       "Las líneas punteadas son el MAE de cada método sobre los diez ensayos")

nota(fig, "Fuente: datos simulados del laboratorio. Cada punto es un experimento fuera de muestra "
          "independiente: el valor real solo se revela después de emitir el pronóstico.")
plt.show()
```

### <span style="color:#8A6100;font-weight:700;letter-spacing:-.2px;">Cómo interpretar el resultado</span>
<div style="border-left:5px solid #B87E00;background:#FFF9E8;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;">Cada fila de la tabla y cada par de puntos de la gráfica es un pequeño experimento fuera de muestra. El origen avanza y el futuro nunca entra antes de tiempo. Un método predecible en su error es más útil para planear que uno que a veces acierta y a veces falla mucho, aunque ambos tuvieran el mismo promedio.</p><p style="margin:0 0 10px 0;">Una advertencia necesaria: la validación <b>no demuestra</b> que el método siempre funcionará. Estima cómo habría funcionado bajo condiciones históricas semejantes. Si el proceso cambia, por una reforma curricular, un cambio en la definición administrativa de deserción o una contingencia, la evidencia pasada deja de aplicar y hay que revalidar.</p></div></div>

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">Anexo A · Extensión voluntaria: corregir el sesgo con una deriva estacional</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>
Las secciones 9 y 11 identificaron una limitación concreta del baseline ganador: subestima **siempre**, porque reproduce la estacionalidad pero ignora la tendencia. Este anexo comprueba esa explicación y la corrige con el cambio más pequeño posible, añadiendo al rezago estacional el incremento anual promedio observado hasta el origen:

$$\hat{y}_{t} = y_{t-2} + \delta_t, \qquad \delta_t=\frac{1}{|H_t|}\sum_{i \in H_t}\left(y_i-y_{i-2}\right)$$

donde $H_t$ contiene únicamente los periodos anteriores a $t$. La deriva $\delta_t$ **se recalcula en cada origen** con la historia disponible en ese momento, de modo que la evaluación sigue libre de fuga de información. La idea no es nueva: combina el baseline estacional con el *método de deriva* que describe el capítulo 5.2 de Hyndman y Athanasopoulos [6], consistente en dejar que el pronóstico crezca al ritmo promedio observado en la historia. Una panorámica del repertorio disponible mucho más allá de estos baselines está en la revisión colectiva de Petropoulos y otros [12].

```python
filas_deriva = []

for i in range(inicio, len(df)):
    historia = df.loc[:i - 1, "Deserciones"]              # solo el pasado del origen
    delta    = historia.diff(2).dropna().mean()           # incremento anual promedio
    pred     = df.loc[i - 2, "Deserciones"] + delta
    filas_deriva.append({
        "Periodo_pronosticado": df.loc[i, "Periodo"],
        "Real":                 df.loc[i, "Deserciones"],
        "Estacional":           df.loc[i - 2, "Deserciones"],
        "Deriva_estimada":      delta,
        "Estacional_deriva":    pred,
        "Abs_estacional":       abs(df.loc[i, "Deserciones"] - df.loc[i - 2, "Deserciones"]),
        "Abs_deriva":           abs(df.loc[i, "Deserciones"] - pred),
    })

deriva = pd.DataFrame(filas_deriva)
print(deriva.round(2).to_string(index=False))

mae_deriva = deriva["Abs_deriva"].mean()
print(f"\nMAE rolling-origin · naïve             : {mae_roll_naive:.2f} estudiantes")
print(f"MAE rolling-origin · estacional        : {mae_roll_estacional:.2f} estudiantes")
print(f"MAE rolling-origin · estacional+deriva : {mae_deriva:.2f} estudiantes")
print(f"Reducción respecto del estacional      : {1 - mae_deriva / mae_roll_estacional:.1%}"
      f"   (magnitud inflada por tratarse de una serie simulada casi lineal)")

delta_final = df["Deserciones"].diff(2).dropna().mean()
pron_deriva = df.iloc[-2]["Deserciones"] + delta_final
print(f"\nDeriva estimada con toda la historia   : {delta_final:+.2f} estudiantes por año")
print(f"Pronóstico estacional+deriva 2027-1    : {pron_deriva:.2f}  (frente a {pronostico_2027_1_estacional} del estacional puro)")
print(f"Capacidad con el mismo criterio        : {int(np.ceil(pron_deriva + mae_deriva))} estudiantes")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.6, 4.4),
                               gridspec_kw={"width_ratios": [1.9, 1]})
x = np.arange(len(deriva))

ax1.plot(x, deriva["Real"], marker="o", linewidth=2.8, color=COLOR["Real"],
         markerfacecolor="#FFFFFF", markeredgecolor=COLOR["Real"], markeredgewidth=1.9,
         label="Real", zorder=4)
ax1.plot(x, deriva["Estacional"], marker="^", linestyle="--", linewidth=2.0,
         color=COLOR["Estacional"], label="Estacional", zorder=3)
ax1.plot(x, deriva["Estacional_deriva"], marker="D", markersize=5, linestyle="-.", linewidth=2.0,
         color=COLOR["Deriva"], label="Estacional + deriva", zorder=3)
ax1.set_xticks(x)
ax1.set_xticklabels(deriva["Periodo_pronosticado"], rotation=45, ha="right")
ax1.set_ylabel("Estudiantes por semestre")
ax1.set_ylim(48, 74)
ax1.legend(loc="upper left", ncols=3)
titulo(ax1, "A · La deriva cierra la brecha de nivel",
       "Diez orígenes móviles · la deriva se reestima en cada uno con solo el pasado")

barras = ax2.bar(["Naïve", "Estacional", "Estacional\n+ deriva"],
                 [mae_roll_naive, mae_roll_estacional, mae_deriva],
                 color=[COLOR["Naive"], COLOR["Estacional"], COLOR["Deriva"]], width=0.58, zorder=3)
for barra, valor in zip(barras, [mae_roll_naive, mae_roll_estacional, mae_deriva]):
    ax2.text(barra.get_x() + barra.get_width() / 2, valor + 0.12, f"{valor:.2f}",
             ha="center", va="bottom", fontsize=10.5, fontweight="bold", color=barra.get_facecolor())
ax2.set_ylabel("MAE rolling-origin · estudiantes")
ax2.set_ylim(0, 5.4)
ax2.grid(axis="x", visible=False)
titulo(ax2, "B · MAE en los diez orígenes",
       "Menor es mejor · la caída es tan grande porque la serie es simulada")

nota(fig, "Fuente: datos simulados del laboratorio. La deriva de cada origen se estima únicamente con los "
          "periodos anteriores a él, de modo que la comparación permanece libre de fuga de información.")
plt.show()
```

La explicación era correcta, y por eso la corrección funciona.</p><p style="margin:0 0 10px 0;">Tres advertencias, para no concluir de más:</p><ul style="margin:0;padding-left:20px;"><li style="margin-bottom:6px;">Buena parte de esa mejora se debe a que la serie es <b>simulada</b> y su tendencia es casi perfectamente lineal. En datos institucionales reales el ruido sería mayor y la ganancia, más modesta.</li><li style="margin-bottom:6px;">Añadir la deriva agrega un parámetro estimado, y con 24 observaciones eso también agrega varianza. Un método con más piezas no es automáticamente mejor: lo es cuando la validación fuera de muestra lo confirma, como aquí.</li><li style="margin-bottom:6px;">La corrección resuelve el sesgo, no la incertidumbre. Sigue siendo necesario un margen de decisión, aunque ahora sea mucho más pequeño.</li></ul><p style="margin:0 0 10px 0;">Con el mismo criterio de decisión de la sección 13, pronóstico más el MAE redondeado hacia arriba, este método da un pronóstico de 64.05 para 2027-1 y una capacidad sugerida de 65 estudiantes: <b>la misma decisión operativa</b> a la que llevó el baseline estacional con un margen mayor. Que dos rutas independientes lleguen al mismo número refuerza la recomendación, y muestra que un margen bien calibrado puede compensar un método más simple.</p></div></div>
