## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">1 · Preparación del entorno</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Usaremos `numpy` para algunos cálculos, `pandas` para organizar los datos [1] y `matplotlib` para construir las gráficas [2].

El código está completo. El objetivo no es memorizar instrucciones de Python, sino relacionar cada resultado con la idea estadística que representa.

```python
# ─────────────────────────────────────────────────────────────────────────
# Actividad Tema 2 · MIAAD · UACJ
# Entorno, identidad visual institucional y utilidades de figura
# ─────────────────────────────────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

# --- Configuración indicada en la consigna, sin cambios -------------------
pd.set_option("display.float_format", lambda value: f"{value:.3f}")
plt.style.use("seaborn-v0_8-whitegrid")

# --- Paleta institucional UACJ --------------------------------------------
# Azul y oro provienen del escudo. Los tonos de serie se eligieron con
# contraste suficiente sobre fondo blanco para que las curvas sigan
# distinguiéndose impresas en escala de grises.
UACJ = {
    "azul":      "#003DA6",   # azul del escudo, títulos y acentos
    "oro":       "#FFEA00",   # oro del escudo, subrayados
    "real":      "#1157CC",   # serie observada
    "sma":       "#B87E00",   # media móvil simple
    "wma":       "#1B7A3E",   # media móvil ponderada
    "ema":       "#A8408A",   # suavizado exponencial
    "ingenuo":   "#C0392B",   # referencia ingenua
    "estacional": "#5A6572",  # referencia estacional de periodo 2
    "deriva":    "#0E7C7B",   # corrección de retraso (anexo A)
    "tinta":     "#1F2933",
    "tinta_sec": "#5A6572",
    "rejilla":   "#DCE1E7",
}

# Un color por método, no por posición: el mismo método conserva su color en
# todas las figuras del cuaderno.
COLOR = {
    "Real":         UACJ["real"],
    "SMA":          UACJ["sma"],
    "WMA":          UACJ["wma"],
    "EMA":          UACJ["ema"],
    "Ingenuo":      UACJ["ingenuo"],
    "Estacional_2": UACJ["estacional"],
    "Estacional_3": UACJ["estacional"],
    "SMA_deriva":   UACJ["deriva"],
}

# El nombre impreso en las figuras no es la clave del diccionario.
NOMBRE = {
    "Real":         "Real",
    "SMA":          "SMA",
    "WMA":          "WMA",
    "EMA":          "EMA",
    "Ingenuo":      "Ingenuo",
    "Estacional_2": "Estacional 2",
    "Estacional_3": "Estacional 3",
    "SMA_deriva":   "SMA + deriva",
}

mpl.rcParams.update({
    "figure.figsize":      (10.6, 4.4),
    "figure.dpi":          105,
    "figure.facecolor":    "#FFFFFF",
    "savefig.facecolor":   "#FFFFFF",
    "axes.facecolor":      "#FFFFFF",
    "axes.edgecolor":      UACJ["rejilla"],
    "axes.linewidth":      1.0,
    "axes.labelcolor":     UACJ["tinta_sec"],
    "axes.titlecolor":     UACJ["azul"],
    "axes.titlesize":      12.5,
    "axes.titleweight":    "bold",
    "axes.titlelocation":  "left",
    "axes.titlepad":       12,
    "axes.labelsize":      10,
    "axes.spines.top":     False,
    "axes.spines.right":   False,
    "axes.grid":           True,
    "axes.axisbelow":      True,
    "grid.color":          UACJ["rejilla"],
    "grid.linewidth":      0.8,
    "text.color":          UACJ["tinta"],
    "xtick.color":         UACJ["tinta_sec"],
    "ytick.color":         UACJ["tinta_sec"],
    "xtick.labelsize":     9,
    "ytick.labelsize":     9,
    "legend.frameon":      False,
    "legend.fontsize":     9.5,
    "lines.linewidth":     2.2,
    "lines.markersize":    5.5,
    "font.family":         "sans-serif",
    "font.sans-serif":     ["DejaVu Sans", "Helvetica", "Arial"],
    "font.size":           10,
})


def titulo(ax, principal, secundario=None):
    """Título en azul institucional con una línea de contexto en tinta secundaria."""
    ax.set_title(principal, loc="left", color=UACJ["azul"], fontsize=12.5,
                 fontweight="bold", pad=26 if secundario else 12)
    if secundario:
        ax.text(0, 1.035, secundario, transform=ax.transAxes, fontsize=9.2,
                color=UACJ["tinta_sec"], ha="left", va="bottom")


def eje_semanal(ax, indice, etiqueta="Semana"):
    """Una marca por semana: el eje temporal no debe interpolarse a ojo."""
    ax.set_xticks(np.asarray(indice))
    ax.set_xlabel(etiqueta)


def rotula(ax, x, y, texto, color, dx=0.28, dy=0.0, ha="left", peso="bold",
           tam=9.5):
    """Etiqueta junto a un punto, con halo blanco para que no se pierda."""
    t = ax.text(x + dx, y + dy, texto, color=color, fontsize=tam,
                fontweight=peso, ha=ha, va="center")
    t.set_path_effects([pe.withStroke(linewidth=3.2, foreground="#FFFFFF")])
    return t


def etiquetas_finales(ax, tabla, columnas, dx=0.22, separacion=1.9):
    """Rótulo al final de cada curva, separando los que se encimarían.

    Con cuatro series sobre doce puntos, una leyenda obliga a ir y volver
    entre el recuadro y la curva; el rótulo directo evita ese salto.
    """
    puntos = []
    for columna in columnas:
        serie_columna = tabla[columna].dropna()
        if len(serie_columna):
            puntos.append([serie_columna.index[-1], float(serie_columna.iloc[-1]),
                           columna])
    puntos.sort(key=lambda p: p[1])
    for i in range(1, len(puntos)):
        if puntos[i][1] - puntos[i - 1][1] < separacion:
            puntos[i][1] = puntos[i - 1][1] + separacion
    for x, y, columna in puntos:
        rotula(ax, x, y, NOMBRE[columna], COLOR[columna], dx=dx)


def nota(fig, texto, alto=0.055):
    """Cierra la figura: reserva una franja inferior y escribe ahí la fuente.

    El pie va dentro del lienzo, no en coordenadas negativas: así sobrevive a
    un savefig normal y no depende de que el backend recorte con bbox_inches.
    """
    fig.tight_layout(rect=(0, alto, 1, 1))
    fig.text(0.005, alto / 2, texto, fontsize=8.2, color=UACJ["tinta_sec"],
             ha="left", va="center")


FUENTE = ("Fuente: datos didácticos de la actividad. No corresponden a "
          "registros de una oficina real.")

print(f"Entorno listo · pandas {pd.__version__} · numpy {np.__version__} "
      f"· matplotlib {mpl.__version__}")
print(f"Paleta institucional cargada: {len(UACJ)} colores · "
      f"azul {UACJ['azul']} y oro {UACJ['oro']} del escudo")
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">2 · El caso: solicitudes ciudadanas por semana</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Durante doce semanas, una oficina registró el número de solicitudes recibidas. La administración quiere responder dos preguntas distintas:

- **Descripción:** ¿cuál es la dirección general de la demanda, más allá de las subidas y bajadas semanales?
- **Pronóstico:** ¿cuántas solicitudes podrían esperarse la semana siguiente para organizar al personal?

Antes de calcular un promedio, debemos mirar la serie original. Una gráfica no es solo una figura: representa una historia que ocurre en orden temporal.

```python
datos = [40, 44, 46, 52, 48, 56, 60, 58, 64, 68, 66, 74]

serie = pd.Series(datos, index=np.arange(1, len(datos) + 1), name="Solicitudes observadas")
serie.index.name = "Semana"
display(serie.to_frame())

# --- Figura 1 · la serie observada, todavía sin ningún método -------------
fig, ax = plt.subplots()
ax.plot(serie.index, serie.to_numpy(), marker="o", color=COLOR["Real"],
        markerfacecolor="#FFFFFF", markeredgecolor=COLOR["Real"],
        markeredgewidth=1.8, zorder=3)

# Las tres semanas que se mueven en contra de la dirección general.
retrocesos = serie.index[1:][np.diff(serie.to_numpy()) < 0]
ax.scatter(retrocesos, serie.loc[retrocesos], s=118, marker="v",
           color=UACJ["ingenuo"], zorder=4)

eje_semanal(ax, serie.index)
ax.set_ylabel("Número de solicitudes")
ax.set_ylim(34, 82)
ax.set_xlim(0.4, 12.9)
titulo(ax, "Solicitudes recibidas por semana",
       "Doce semanas observadas · antes de aplicar cualquier método de suavizado")

rotula(ax, serie.index[0], serie.iloc[0], "40", COLOR["Real"], dx=0.0, dy=-2.6,
       ha="center")
rotula(ax, serie.index[-1], serie.iloc[-1], "74", COLOR["Real"], dx=0.0, dy=2.6,
       ha="center")
ax.annotate("las tres semanas que bajan\ncaen en la 5, la 8 y la 11",
            xy=(8, 58), xytext=(3.1, 71.5), fontsize=9.2,
            color=UACJ["ingenuo"], ha="left",
            arrowprops=dict(arrowstyle="->", color=UACJ["ingenuo"], lw=1.1,
                            connectionstyle="arc3,rad=-0.22"))
ax.annotate(f"+{serie.iloc[-1] - serie.iloc[0]:.0f} solicitudes en once semanas\n"
            f"({(serie.iloc[-1] - serie.iloc[0]) / 11:.2f} por semana en promedio)",
            xy=(11.6, 41.5), fontsize=9.2, color=UACJ["tinta_sec"], ha="right")

nota(fig, FUENTE)
plt.show()

# Los dos resúmenes de la dirección general, y los extremos de cada mitad, se
# imprimen para que ninguna cifra de la interpretación quede afirmada sin
# haberse calculado aquí.
print(f"Avance medio entre semanas consecutivas : {serie.diff().dropna().mean():.3f}")
print(f"Pendiente por mínimos cuadrados         : "
      f"{np.polyfit(serie.index, serie.to_numpy(), 1)[0]:.3f} solicitudes por semana")
print(f"Máximo de las semanas 1 a 6             : {serie.loc[1:6].max():.0f} "
      f"(semana {serie.loc[1:6].idxmax()})")
print(f"Mínimo de las semanas 7 a 12            : {serie.loc[7:12].min():.0f} "
      f"(semana {serie.loc[7:12].idxmin()})")
```

### <span style="color:#8A6100;font-weight:700;letter-spacing:-.2px;">Cómo interpretar el resultado</span>

<div style="border-left:5px solid #B87E00;background:#FFF9E8;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;">La tabla tiene una sola columna de valores, pero dos elementos inseparables: <b>el número de solicitudes</b> y <b>la semana en que se registró</b>. Separarlos destruye la serie; un conteo sin semana es un número suelto, no una observación temporal.</p><p style="margin:0 0 10px 0;">En la figura conviene mirar tres cosas y en este orden. ## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">3 · Tres maneras de resumir el pasado</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Aplicaremos los tres métodos con parámetros sencillos:

- **SMA con $k=3$:** cada valor es el promedio de tres semanas y todas tienen la misma importancia.
- **WMA con pesos $[0.2, 0.3, 0.5]$:** también usa tres semanas, pero da mayor importancia a la más reciente. Los pesos se escriben de la semana más antigua a la más reciente.
- **EMA con $\alpha=0.3$:** combina el dato actual con el nivel suavizado anterior. Un $\alpha$ pequeño conserva más memoria; uno grande reacciona más al dato reciente.

Estos parámetros no son universales. Por ahora los fijaremos para observar con claridad cómo cambia el comportamiento de cada método.

Los tres son casos particulares de una misma idea, el promedio móvil, descrita en la bibliografía del curso [3]; la forma recursiva de la EMA proviene del trabajo de Brown en control de inventarios [4].

```python
k = 3
pesos = np.array([0.2, 0.3, 0.5])
alpha = 0.3

print("Suma de los pesos WMA:", pesos.sum())

# --- Figura 2 · cuánto pesa cada semana en el valor suavizado -------------
# Un método de suavizado es un reparto de pesos sobre el pasado. Dibujarlo
# hace visible de antemano cuál reacciona más y cuál mira más atrás.
rezagos = np.arange(6)
peso_sma = np.where(rezagos < k, 1 / k, 0.0)
peso_wma = np.array([0.5, 0.3, 0.2, 0.0, 0.0, 0.0])      # del más reciente al más antiguo
peso_ema = alpha * (1 - alpha) ** rezagos

retraso_sma = float((peso_sma * rezagos).sum())
retraso_wma = float((peso_wma * rezagos).sum())
retraso_ema = (1 - alpha) / alpha                         # límite teórico de la serie geométrica

fig, ax = plt.subplots(figsize=(10.6, 4.0))
ancho = 0.26
for desplazamiento, valores, clave, retraso in [
    (-ancho, peso_sma, "SMA", retraso_sma),
    (0.0,    peso_wma, "WMA", retraso_wma),
    (ancho,  peso_ema, "EMA", retraso_ema),
]:
    ax.bar(rezagos + desplazamiento, valores, width=ancho, color=COLOR[clave],
           label=f"{clave} · retraso medio {retraso:.2f}", zorder=3)

ax.set_xticks(rezagos)
ax.set_xticklabels(["la semana\nmás reciente", "1 antes", "2 antes", "3 antes",
                    "4 antes", "5 antes"])
ax.set_ylabel("Peso asignado")
ax.set_ylim(0, 0.58)
ax.legend(loc="upper right", frameon=False)
titulo(ax, "Cuánto pesa cada semana dentro del valor suavizado",
       "SMA reparte por igual, WMA carga la más reciente y EMA nunca olvida del todo")
ax.annotate("la EMA sigue asignando peso\na semanas que SMA y WMA ya descartaron",
            xy=(4, 0.075), xytext=(2.55, 0.30), fontsize=9.2,
            color=UACJ["ema"], ha="left",
            arrowprops=dict(arrowstyle="->", color=UACJ["ema"], lw=1.1,
                            connectionstyle="arc3,rad=0.2"))

nota(fig, "Pesos implícitos de cada método con k = 3, pesos [0.2, 0.3, 0.5] y α = 0.3.")
plt.show()

print(f"\nRetraso medio del valor suavizado respecto de la semana actual:")
print(f"  SMA  {retraso_sma:.2f} semanas   ·  (k - 1) / 2")
print(f"  WMA  {retraso_wma:.2f} semanas   ·  0(0.5) + 1(0.3) + 2(0.2)")
print(f"  EMA  {retraso_ema:.2f} semanas   ·  (1 - α) / α")
```

### <span style="color:#8A6100;font-weight:700;letter-spacing:-.2px;">Cómo interpretar el resultado</span>

<div style="border-left:5px solid #B87E00;background:#FFF9E8;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;">La figura muestra los tres métodos como lo que realmente son: <b>tres maneras de repartir peso sobre el pasado</b>. El SMA da un tercio a cada una de las tres últimas semanas y cero a lo demás. El WMA usa las mismas tres semanas pero carga la mitad del peso en la más reciente. La EMA reparte peso <i>α</i>(1&nbsp;−&nbsp;<i>α</i>)<sup><i>j</i></sup> sobre <i>todas</i> las semanas anteriores: nunca olvida del todo, solo va olvidando poco a poco.</p><p style="margin:0 0 10px 0;">De ese reparto sale el número que gobierna todo el resto del cuaderno: el <b>retraso medio</b>, es decir cuántas semanas atrás está, en promedio, la información que el método resume. Vale 1.00 semanas para el SMA, 0.70 para el WMA y 2.33 para la EMA con α = 0.3.</p><p style="margin:0 0 10px 0;">El retraso no es un defecto de programación: es el precio de promediar. Cuando la serie no tiene dirección, ese precio es cercano a cero. Cuando la serie sube de manera sostenida, cada semana de retraso se paga en solicitudes que el pronóstico deja de contar. La sección 6 medirá exactamente cuántas.</p></div></div>

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">4 · Primero: suavizar la serie</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Un **valor suavizado de la semana $t$** resume la información disponible hasta esa misma semana, incluido el dato observado en $t$. Sirve para reducir las variaciones cortas y apreciar mejor el nivel o la dirección general.

Por ejemplo, el SMA suavizado de la semana 3 usa las semanas 1, 2 y 3. Como ya conoce la semana 3, **no es todavía un pronóstico de esa semana**.

El suavizado exponencial simple, del que la EMA de esta actividad es la forma recursiva, se desarrolla en Hyndman y Athanasopoulos [5], sección 8.1; los dos métodos de referencia que aparecerán en la sección 6 están en su sección 5.2.

```python
suavizados = pd.DataFrame({"Real": serie})

# SMA: promedio de una ventana de tamaño k
suavizados["SMA"] = serie.rolling(window=k).mean()

# WMA: producto punto entre la ventana y los pesos
suavizados["WMA"] = serie.rolling(window=len(pesos)).apply(
    lambda ventana: np.dot(ventana, pesos), raw=True
)

# EMA: actualización recursiva
suavizados["EMA"] = serie.ewm(alpha=alpha, adjust=False).mean()

display(suavizados.round(3))
```

### Verificación manual: ¿qué entra en cada cálculo?

Para la **semana 3**, el SMA suavizado utiliza los datos 40, 44 y 46:

$$
SMA_3=\frac{40+44+46}{3}=43.33
$$

El WMA usa las mismas semanas, pero otorga más peso a la más reciente:

$$
WMA_3=0.2(40)+0.3(44)+0.5(46)=44.20
$$

Para la **semana 4**, la ventana avanza: sale la semana 1 y entran las semanas 2, 3 y 4. Así, $SMA_4=(44+46+52)/3=47.33$.

```python
print("SMA suavizado de la semana 3:", suavizados.loc[3, "SMA"])
print("WMA suavizado de la semana 3:", suavizados.loc[3, "WMA"])
print("SMA suavizado de la semana 4:", suavizados.loc[4, "SMA"])
print("EMA suavizado de la semana 4:", suavizados.loc[4, "EMA"])
```

```python
# --- Figura 3 · la serie y sus tres versiones suavizadas ------------------
fig, ax = plt.subplots()

# Franja donde SMA y WMA todavía no existen: la ventana no está completa.
ax.axvspan(0.4, 2.5, color=UACJ["rejilla"], alpha=0.45, zorder=0)
ax.text(1.45, 71.5, "sin SMA ni WMA:\nla ventana de tres\naún no se llena",
        fontsize=8.8, color=UACJ["tinta_sec"], ha="center", va="center")

ax.plot(suavizados.index, suavizados["Real"], marker="o", color=COLOR["Real"],
        markerfacecolor="#FFFFFF", markeredgecolor=COLOR["Real"],
        markeredgewidth=1.8, zorder=5)
for columna in ["SMA", "WMA", "EMA"]:
    ax.plot(suavizados.index, suavizados[columna], marker="o", markersize=4.2,
            color=COLOR[columna], linewidth=2.0, zorder=4)

eje_semanal(ax, suavizados.index)
ax.set_ylabel("Solicitudes")
ax.set_ylim(34, 82)
ax.set_xlim(0.4, 13.9)
titulo(ax, "Datos observados y valores suavizados",
       "Cada curva suavizada usa el dato de su propia semana: todavía no es un pronóstico")
etiquetas_finales(ax, suavizados, ["Real", "SMA", "WMA", "EMA"])

ax.annotate("la caída de la semana 5 casi desaparece\nen las curvas suavizadas",
            xy=(5, 48), xytext=(5.6, 38.0), fontsize=9.2,
            color=UACJ["tinta_sec"], ha="left",
            arrowprops=dict(arrowstyle="->", color=UACJ["tinta_sec"], lw=1.1,
                            connectionstyle="arc3,rad=0.25"))

nota(fig, FUENTE)
plt.show()
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">5 · Después: convertir el suavizado en pronóstico</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Para pronosticar la semana $t$, el dato real de esa semana aún no existe. El cálculo solamente puede usar información conocida hasta $t-1$.

La diferencia se aprecia en la semana 4:

- **SMA suavizado de la semana 4:** promedia las semanas 2, 3 y 4; incluye el valor 52 de la propia semana 4.
- **Pronóstico SMA para la semana 4:** promedia las semanas 1, 2 y 3; no utiliza el valor 52 que intenta anticipar.

En el código, `.shift(1)` mueve los valores suavizados una semana hacia la derecha para convertirlos en pronósticos de un paso adelante.

```python
pronosticos = pd.DataFrame({"Real": serie})
pronosticos["SMA"] = suavizados["SMA"].shift(1)
pronosticos["WMA"] = suavizados["WMA"].shift(1)
pronosticos["EMA"] = suavizados["EMA"].shift(1)

# Baseline ingenuo: repetir el último dato observado
pronosticos["Ingenuo"] = serie.shift(1)

# Baseline estacional de periodo 2: repetir el dato de dos periodos atrás
pronosticos["Estacional_2"] = serie.shift(2)

display(pronosticos.round(3))
```

### Comprobación del pronóstico para la semana 4

$$
\widehat y_{4|3}=\frac{40+44+46}{3}=43.33
$$

El dato observado de la semana 4 fue 52. Por tanto, el error del pronóstico SMA es $52-43.33=8.67$ solicitudes. El error solo puede calcularse después de conocer el valor real.

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">6 · Comparación justa mediante métricas</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

No basta con observar qué línea parece más cercana. Compararemos todos los métodos solamente en las semanas donde **todos** tienen un pronóstico disponible.

- **MAE:** distancia absoluta promedio entre lo observado y lo pronosticado. Se expresa en solicitudes y es fácil de interpretar.
- **RMSE:** también se expresa en solicitudes, pero aumenta más cuando existen errores grandes.
- **sMAPE:** expresa el error relativo en porcentaje y trata de forma simétrica observados y pronósticos.

Como referencia incluiremos dos métodos simples:

- **Ingenuo:** pronostica que la siguiente semana será igual a la última observada.
- **Estacional de periodo 2:** repite el valor observado dos semanas antes. Aquí es una referencia didáctica; no afirmamos todavía que exista estacionalidad real.

Restringir la comparación a una ventana común no es un tecnicismo: si cada método se evaluara en las semanas que le convienen, la comparación mediría qué método empieza antes, no cuál acierta más [6]. Las tres métricas tienen límites conocidos, y el sMAPE uno que aquí importa: **no es tan simétrico como su nombre indica**. Para un mismo valor observado penaliza más los pronósticos bajos que los altos [7], y en esta serie todos los métodos se quedan cortos. La implementación de la consigna usa valores absolutos en el denominador, con lo que evita el otro problema que señalan Hyndman y Koehler: que el sMAPE llegue a tomar valores negativos.

```python
def calcular_metricas(real, predicho):
    real = np.asarray(real, dtype=float)
    predicho = np.asarray(predicho, dtype=float)

    mae = np.mean(np.abs(real - predicho))
    rmse = np.sqrt(np.mean((real - predicho) ** 2))

    denominador = np.abs(real) + np.abs(predicho)
    smape = np.mean(
        np.where(denominador == 0, 0, 2 * np.abs(real - predicho) / denominador)
    ) * 100

    return {"MAE": mae, "RMSE": rmse, "sMAPE": smape}


def evaluar_en_ventana_comun(tabla, metodos):
    mascara_comun = tabla[["Real"] + metodos].notna().all(axis=1)
    evaluacion = tabla.loc[mascara_comun]

    resultados = {
        metodo: calcular_metricas(evaluacion["Real"], evaluacion[metodo])
        for metodo in metodos
    }

    return pd.DataFrame(resultados).T.sort_values("RMSE"), evaluacion
```

```python
metodos = ["SMA", "WMA", "EMA", "Ingenuo", "Estacional_2"]
metricas_serie, ventana_comun = evaluar_en_ventana_comun(pronosticos, metodos)

print("Primera semana de la ventana común:", ventana_comun.index.min())
print("Número de pronósticos comparados:", len(ventana_comun))
display(metricas_serie.round(3))

# Pronósticos para la semana 13. Todavía no se conoce su valor real.
pronostico_semana_13 = pd.Series({
    "SMA": serie.iloc[-k:].mean(),
    "WMA": np.dot(serie.iloc[-len(pesos):], pesos),
    "EMA": suavizados.loc[12, "EMA"],
    "Ingenuo": serie.loc[12],
    "Estacional_2": serie.loc[11],
}, name="Pronóstico semana 13")

print("\nPronósticos disponibles para planear la semana 13:")
display(pronostico_semana_13.to_frame().round(3))

# --- Error medio con signo: la magnitud no dice hacia dónde se falla ------
sesgo = pd.Series(
    {metodo: (ventana_comun["Real"] - ventana_comun[metodo]).mean()
     for metodo in metodos},
    name="Error medio (real − pronóstico)"
).reindex(metricas_serie.index)

# Va como tabla, no solo dentro de la figura: el argumento de esta sección
# descansa en estas cinco cifras y deben poder leerse sin interpretar barras.
print("\nError medio con signo en la ventana común (positivo = se queda corto):")
display(sesgo.to_frame().round(3))

# --- Figura 5 · exactitud y dirección del error, lado a lado --------------
fig, (izq, der) = plt.subplots(1, 2, figsize=(11.4, 4.2))
posiciones = np.arange(len(metricas_serie))[::-1]

izq.barh(posiciones, metricas_serie["RMSE"],
         color=[COLOR[m] for m in metricas_serie.index], zorder=3, height=0.62)
for y, (metodo, valor) in zip(posiciones, metricas_serie["RMSE"].items()):
    izq.text(valor + 0.14, y, f"{valor:.2f}", va="center", fontsize=9.2,
             color=COLOR[metodo], fontweight="bold")
izq.set_yticks(posiciones)
izq.set_yticklabels([NOMBRE[m] for m in metricas_serie.index])
izq.set_xlim(0, metricas_serie["RMSE"].max() * 1.22)
izq.set_xlabel("RMSE, en solicitudes")
izq.grid(axis="y", visible=False)
titulo(izq, "Exactitud", "Menor es mejor")

der.barh(posiciones, sesgo.to_numpy(),
         color=[COLOR[m] for m in sesgo.index], zorder=3, height=0.62)
for y, (metodo, valor) in zip(posiciones, sesgo.items()):
    der.text(valor + 0.2, y, f"+{valor:.2f}", va="center", fontsize=9.2,
             color=COLOR[metodo], fontweight="bold")
der.axvline(0, color=UACJ["tinta_sec"], linewidth=1.0, zorder=4)
der.set_yticks(posiciones)
der.set_yticklabels([])
der.set_xlim(0, sesgo.max() * 1.3)
der.set_xlabel("Error medio con signo, en solicitudes")
der.grid(axis="y", visible=False)
titulo(der, "Dirección del error",
       "Los cinco valores son positivos: los cinco métodos se quedan cortos")

nota(fig, "Ventana común de nueve semanas, de la 4 a la 12. " + FUENTE, alto=0.075)
plt.show()
```

```python
def construir_pronosticos(valores, k=3, pesos=np.array([0.2, 0.3, 0.5]), alpha=0.3):
    y = pd.Series(valores, index=np.arange(1, len(valores) + 1), dtype=float)
    tabla = pd.DataFrame({"Real": y})

    sma = y.rolling(k).mean()
    wma = y.rolling(len(pesos)).apply(lambda ventana: np.dot(ventana, pesos), raw=True)
    ema = y.ewm(alpha=alpha, adjust=False).mean()

    tabla["SMA"] = sma.shift(1)
    tabla["WMA"] = wma.shift(1)
    tabla["EMA"] = ema.shift(1)
    tabla["Ingenuo"] = y.shift(1)
    tabla["Estacional_2"] = y.shift(2)
    return tabla


resumen = []
tablas = {}

for nombre, valores in series.items():
    tabla = construir_pronosticos(valores, k=k, pesos=pesos, alpha=alpha)
    metricas, evaluacion = evaluar_en_ventana_comun(tabla, metodos)
    tablas[nombre] = tabla

    for metodo, fila in metricas.iterrows():
        resumen.append({"Serie": nombre, "Método": metodo, **fila.to_dict()})

resumen_metricas = pd.DataFrame(resumen)
tabla_rmse = resumen_metricas.pivot(index="Serie", columns="Método", values="RMSE")
display(tabla_rmse.round(3))

# --- Figura 7 · mapa del RMSE: dónde gana cada método --------------------
# El color se normaliza dentro de cada renglón porque las cuatro series
# tienen escalas distintas; el número impreso es siempre el RMSE real.
orden = ["SMA", "WMA", "EMA", "Ingenuo", "Estacional_2"]
mapa = tabla_rmse[orden]
crudo = mapa.to_numpy()
rango = crudo.max(axis=1, keepdims=True) - crudo.min(axis=1, keepdims=True)
normalizado = (crudo - crudo.min(axis=1, keepdims=True)) / np.where(rango == 0, 1, rango)

escala = mpl.colors.LinearSegmentedColormap.from_list(
    "uacj", ["#FFFFFF", "#BBD1F0", "#4E82D6", UACJ["azul"]])

fig, ax = plt.subplots(figsize=(10.6, 3.9))
ax.imshow(normalizado, cmap=escala, aspect="auto", vmin=0, vmax=1)
ax.grid(False)

for i in range(mapa.shape[0]):
    ganador = int(np.argmin(crudo[i]))
    for j in range(mapa.shape[1]):
        ax.text(j, i, f"{crudo[i, j]:.2f}", ha="center", va="center",
                fontsize=10.5, fontweight="bold" if j == ganador else "normal",
                color="#FFFFFF" if normalizado[i, j] > 0.55 else UACJ["tinta"])
    ax.add_patch(mpl.patches.Rectangle((ganador - 0.5, i - 0.5), 1, 1, fill=False,
                                       edgecolor=UACJ["ingenuo"], linewidth=2.4))

ax.set_xticks(range(len(orden)))
ax.set_xticklabels([NOMBRE[m] for m in orden], fontsize=10)
ax.set_yticks(range(mapa.shape[0]))
ax.set_yticklabels([s.replace(" — ", "\n") for s in mapa.index], fontsize=9.2)
ax.tick_params(length=0)
for lado in ax.spines.values():
    lado.set_visible(False)
titulo(ax, "RMSE de cada método en cada serie",
       "Recuadro rojo: el mejor de su renglón · color normalizado por renglón, "
       "el número es el RMSE real")

nota(fig, "Ventana común por serie: todos los métodos se comparan en los mismos "
          "periodos.", alto=0.05)
plt.show()
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">8 · Selección de alpha sin mirar el futuro</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Usaremos la **Serie 2**, cuyo nivel es relativamente estable. Probaremos varios valores de $\alpha$ en los periodos 13–16, que funcionarán como validación. Después evaluaremos una sola vez en los periodos 17–20, reservados como prueba final.

La regla es sencilla: una decisión sobre el parámetro debe tomarse con datos anteriores al periodo que se usará para reportar el desempeño final. De lo contrario, estaríamos usando indirectamente información del futuro.

Esa regla tiene un motivo técnico: en series temporales la partición no puede ser aleatoria, porque mezclaría pasado y futuro dentro del mismo conjunto [9]. La literatura sobre suavizado exponencial recoge además décadas de discusión sobre cómo elegir $\alpha$ y qué tan estable resulta esa elección [10].

```python
y_alpha = pd.Series(series["Serie 2 — nivel estable"], index=np.arange(1, 21), dtype=float)
alphas = np.arange(0.05, 1.00, 0.05)
errores_validacion = []

for candidato in alphas:
    pronostico = y_alpha.ewm(alpha=candidato, adjust=False).mean().shift(1)
    real_val = y_alpha.loc[13:16]
    pred_val = pronostico.loc[13:16]
    rmse_val = calcular_metricas(real_val, pred_val)["RMSE"]
    errores_validacion.append(rmse_val)

mejor_alpha = float(alphas[np.argmin(errores_validacion)])
print(f"Mejor α en validación: {mejor_alpha:.2f}")
print(f"RMSE de validación con ese α: {min(errores_validacion):.4f}")
print(f"RMSE de validación con α = 0.30, el valor fijado al inicio: "
      f"{errores_validacion[list(np.round(alphas, 2)).index(0.30)]:.4f}")

# --- Figura 8 · la curva de validación -----------------------------------
fig, ax = plt.subplots(figsize=(10.0, 4.0))
ax.plot(alphas, errores_validacion, marker="o", color=COLOR["EMA"], zorder=3)
ax.scatter([mejor_alpha], [min(errores_validacion)], s=150, zorder=5,
           color=UACJ["ingenuo"], edgecolor="#FFFFFF", linewidth=1.8)
ax.set_xlabel("α")
ax.set_ylabel("RMSE de validación")
ax.set_xlim(0, 1.0)
titulo(ax, "Selección temporal de α sobre la Serie 2",
       "Validación en los periodos 13 a 16 · la prueba final, 17 a 20, no interviene aquí")
rotula(ax, mejor_alpha, min(errores_validacion),
       f"α = {mejor_alpha:.2f}  ·  RMSE {min(errores_validacion):.3f}",
       UACJ["ingenuo"], dx=0.035, dy=0.03)
ax.annotate("la curva solo sube: el mínimo cae en el borde\n"
            "de la rejilla, no en un punto interior",
            xy=(0.62, errores_validacion[11]), xytext=(0.34, 0.83),
            fontsize=9.2, color=UACJ["tinta_sec"], ha="left",
            arrowprops=dict(arrowstyle="->", color=UACJ["tinta_sec"], lw=1.1,
                            connectionstyle="arc3,rad=-0.2"))

nota(fig, "Cada punto es el RMSE de cuatro pronósticos. La evidencia por "
          "candidato es escasa y conviene decirlo.")
plt.show()
```

```python
pronostico_final = y_alpha.ewm(alpha=mejor_alpha, adjust=False).mean().shift(1)
metricas_prueba = calcular_metricas(y_alpha.loc[17:20], pronostico_final.loc[17:20])

print("Métricas en prueba final (periodos 17–20):")
display(pd.DataFrame([metricas_prueba], index=[f"EMA α={mejor_alpha:.2f}"]).round(3))

# --- Referencias de contraste, evaluadas en los mismos periodos 17–20 -----
# El α elegido solo significa algo comparado con algo. Ninguna de estas
# referencias usó los periodos de prueba para ajustarse.
contraste = {
    f"EMA α={mejor_alpha:.2f} (elegida en validación)": pronostico_final.loc[17:20],
    "EMA α=0.30 (el valor fijado al inicio)":
        y_alpha.ewm(alpha=0.30, adjust=False).mean().shift(1).loc[17:20],
    "Ingenuo": y_alpha.shift(1).loc[17:20],
    "Media de toda la historia hasta el periodo 16":
        pd.Series(y_alpha.loc[:16].mean(), index=np.arange(17, 21)),
}

tabla_contraste = pd.DataFrame(
    {nombre: calcular_metricas(y_alpha.loc[17:20], valores)
     for nombre, valores in contraste.items()}
).T.sort_values("RMSE")

print("\nMismo periodo de prueba, cuatro maneras de pronosticarlo:")
display(tabla_contraste.round(3))

# Las dos constantes de memoria máxima no son la misma, y conviene no
# confundirlas: el límite de la EMA con adjust=False cuando α → 0 es el primer
# dato, no la media histórica, porque la recursión arranca en él.
print("\nDos constantes de memoria máxima sobre la Serie 2:")
print(f"  primer dato observado, límite de la EMA con α → 0 : {y_alpha.loc[1]:.3f}")
print(f"  media de toda la historia hasta el periodo 16      : {y_alpha.loc[:16].mean():.3f}")
print(f"  peso que la EMA con α = {mejor_alpha:.2f} conserva sobre el primer dato "
      f"en el periodo 16: {(1 - mejor_alpha) ** 15:.3f}")
```

## <span style="color:#003DA6;font-weight:800;letter-spacing:-.2px;">Anexo A · Poner a prueba el ciclo de tres semanas y corregir el retraso</span>
<div style="height:4px;background:linear-gradient(90deg,#003DA6 0%,#1157CC 45%,#FFEA00 100%);border-radius:2px;margin:-8px 0 16px 0;"></div>

Dos objetivos que validar aqui:

El primero viene de la pregunta 1: los tres retrocesos de la serie caen en las semanas 5, 8 y 11, separados por tres semanas. Si eso fuera un ciclo aprovechable, una referencia que repita el dato de tres semanas atrás debería competir con las demás. Se pone a prueba en lugar de suponerlo.

El segundo viene de la sección 6. Allí quedó demostrado que los métodos de suavizado no fallan por promediar, sino por el retraso que el promedio introduce cuando la serie tiene tendencia. Si el diagnóstico es correcto, corregir el retraso debería bastar, sin cambiar de método: al pronóstico SMA se le suma el avance medio observado hasta el origen, multiplicado por su retraso, $(k+1)/2$. Es la misma idea que Holt formalizó al añadir una componente de tendencia al suavizado exponencial [11]. El avance se estima en cada semana usando solo datos anteriores, de modo que no hay fuga de información.

```python
# --- A.1 · ¿Existe de verdad un ciclo de tres semanas? -------------------
# La pregunta 1 dejó abierta una sospecha: los tres retrocesos caen en las
# semanas 5, 8 y 11. Si eso fuera un ciclo aprovechable, repetir el dato de
# tres semanas atrás debería competir. Se pone a prueba, no se supone.
anexo = pd.DataFrame({"Real": serie})
anexo["SMA"] = suavizados["SMA"].shift(1)
anexo["Ingenuo"] = serie.shift(1)
anexo["Estacional_3"] = serie.shift(3)

# --- A.2 · Corregir el retraso en lugar de cambiar de método -------------
# Al pronóstico SMA se le suma el avance medio observado *hasta el origen*,
# multiplicado por el retraso teórico del SMA, (k + 1) / 2. En cada semana el
# avance se estima solo con datos anteriores: no hay fuga de información.
retraso_total_sma = (k + 1) / 2
correccion = []
for semana in serie.index:
    historia = serie.loc[:semana - 1]
    if len(historia) < 4 or pd.isna(anexo.loc[semana, "SMA"]):
        correccion.append(np.nan)
        continue
    avance = historia.diff().dropna().mean()
    correccion.append(anexo.loc[semana, "SMA"] + avance * retraso_total_sma)
anexo["SMA_deriva"] = correccion

metodos_anexo = ["SMA", "Ingenuo", "Estacional_3", "SMA_deriva"]
metricas_anexo, ventana_anexo = evaluar_en_ventana_comun(anexo, metodos_anexo)
metricas_anexo["Error medio"] = [
    (ventana_anexo["Real"] - ventana_anexo[m]).mean() for m in metricas_anexo.index
]

print("Ventana común del anexo:", ventana_anexo.index.min(), "a",
      ventana_anexo.index.max(), f"({len(ventana_anexo)} semanas)")
display(metricas_anexo.round(3))

# --- Figura 9 · el retraso corregido, y el ciclo de tres descartado -------
fig, (izq, der) = plt.subplots(1, 2, figsize=(12.2, 4.3),
                               gridspec_kw={"width_ratios": [1.55, 1]})

izq.plot(anexo.index, anexo["Real"], marker="o", color=COLOR["Real"],
         markerfacecolor="#FFFFFF", markeredgecolor=COLOR["Real"],
         markeredgewidth=1.8, zorder=5)
for columna in ["SMA", "SMA_deriva"]:
    izq.plot(anexo.index, anexo[columna], marker="o", markersize=4.2,
             color=COLOR[columna], linewidth=2.0, zorder=4)
izq.fill_between(anexo.index, anexo["SMA"], anexo["Real"], color=COLOR["SMA"],
                 alpha=0.13, zorder=1)
eje_semanal(izq, anexo.index)
izq.set_ylabel("Solicitudes")
izq.set_ylim(34, 82)
izq.set_xlim(0.4, 14.6)
titulo(izq, "El SMA con su retraso corregido",
       "La franja es lo que el SMA pierde por mirar hacia atrás")
etiquetas_finales(izq, anexo, ["Real", "SMA", "SMA_deriva"], separacion=3.4)

posiciones = np.arange(len(metricas_anexo))[::-1]
der.barh(posiciones, metricas_anexo["RMSE"],
         color=[COLOR[m] for m in metricas_anexo.index], zorder=3, height=0.6)
for y, (metodo, fila) in zip(posiciones, metricas_anexo.iterrows()):
    der.text(fila["RMSE"] + 0.16, y,
             f"{fila['RMSE']:.2f}   (sesgo {fila['Error medio']:+.2f})",
             va="center", fontsize=9.0, color=COLOR[metodo], fontweight="bold")
der.set_yticks(posiciones)
der.set_yticklabels([NOMBRE[m] for m in metricas_anexo.index])
der.set_xlim(0, metricas_anexo["RMSE"].max() * 1.75)
der.set_xlabel("RMSE, en solicitudes")
der.grid(axis="y", visible=False)
titulo(der, "Ocho semanas comparables", "Menor es mejor")

nota(fig, "Ventana común 5–12, ocho pronósticos. " + FUENTE, alto=0.075)
plt.show()

avance_final = serie.diff().dropna().mean()
print(f"\nAvance medio observado en las doce semanas: {avance_final:.3f} solicitudes")
print(f"Pronóstico SMA para la semana 13            : {serie.iloc[-k:].mean():.2f}")
print(f"Pronóstico SMA + deriva para la semana 13   : "
      f"{serie.iloc[-k:].mean() + avance_final * retraso_total_sma:.2f}")
print(f"Pronóstico ingenuo para la semana 13        : {serie.loc[12]:.2f}")
print(f"\nTres avances semanales, referencia del sesgo del Estacional 3: "
      f"{3 * avance_final:.2f} solicitudes")
```

### <span style="color:#8A6100;font-weight:700;letter-spacing:-.2px;">Cómo interpretar el resultado</span>

<div style="border-left:5px solid #B87E00;background:#FFF9E8;padding:14px 18px;margin:6px 0 14px 0;border-radius:0 10px 10px 0;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="color:#1F2933;font-size:14.5px;line-height:1.7;"><p style="margin:0 0 10px 0;"><b>El ciclo de tres semanas no existe.</b> La referencia que repite el dato de tres semanas atrás obtiene el peor RMSE del anexo, 8.46, muy por detrás del SMA, 6.02, y del ingenuo, 5.24. La razón es clara: repetir un valor de tres semanas atrás arrastra tres semanas de tendencia sin compensar, y su sesgo de +8.25 solicitudes es del orden de esos tres avances semanales, que sumarían 9.27 si la tendencia fuera perfectamente constante. La coincidencia de las semanas 5, 8 y 11 era eso, una coincidencia, y con doce datos no había cómo distinguirla del ruido. Registrarla como hipótesis y descartarla con evidencia es el procedimiento correcto.</p><p style="margin:0 0 10px 0;"><b>El retraso sí era el problema.</b> El SMA con la deriva añadida obtiene un RMSE de 3.56 frente a 6.02 del SMA sin corregir, y su error medio pasa de +5.42 a −0.54: el sesgo prácticamente desaparece. Es el mismo método, los mismos datos y la misma ventana; lo único que cambió fue reconocer que un promedio de tres semanas está centrado dos semanas atrás y sumar el avance correspondiente.</p><p style="margin:0 0 10px 0;">Y con esa corrección el suavizado sí supera al pronóstico ingenuo, 3.56 contra 5.24, lo que matiza la conclusión de la sección 6: lo que fallaba no era promediar, era promediar sin devolver la tendencia que el promedio se come.</p><p style="margin:0 0 10px 0;">Dos límites que conviene declarar. La ventana común del anexo tiene ocho semanas, una menos que la de la sección 6, porque la deriva necesita historia para estimarse; con ocho observaciones ninguna diferencia es concluyente. Y la corrección supone que el avance se mantiene: si la serie se aplana, el término de deriva empezará a sobreestimar con la misma sistematicidad con que el SMA subestimaba.</p></div></div>
